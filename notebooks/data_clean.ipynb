```python
import pandas as pd 

df = pd.read_csv(r'..\data\raw_data.csv')
```

```python
df.isnull().sum()
```

```python
num_cols = df.select_dtypes(include=['int64','float64']).columns

for col in num_cols:
    if col!='Churned':
       df[col] = df[col].fillna(df[col].median())

cat_cols = df.select_dtypes(include=['str','object']).columns

for col in cat_cols:
    if col!='Churned':
        df[col] = df[col].fillna(df[col].mode()[0])
```

```python
df.isnull().sum()
```

```python
df['Churned'].value_counts()
print(df['Churned'].value_counts(normalize=True) * 100)
```

#### Generate customer ID

```python
df.insert(
    0,
    'CustomerID',
    ['CUST' + str(i).zfill(5) for i in range(1, len(df) + 1)]
)
```

```python
df[['CustomerID','Churned']].head()
```

```python
df.to_csv('../data/cleaned_customers.csv', index=False)
```
